```python
# 0. PREPARACIÓN DE ENTORNOS Y DATOS DE PRUEBA
# ==========================================
# Importamos las librerías necesarias
import pandas as pd
import numpy as np

# Creamos datos ficticios para simular tus archivos de entrada
# Datos de transacciones (CSV)
datos_transacciones = {
    'id_producto': [1, 2, 3, 4, 5, 6],
    'cantidad': [2, 1, 5, 3, None, 1],  # Incluye un valor nulo para el ejemplo
    'fecha': ['2026-08-01', '2026-08-02', '2026-08-02', '2026-08-03', '2026-08-04', '2026-08-05']
}
df_transacciones_dummy = pd.DataFrame(datos_transacciones)
df_transacciones_dummy.to_csv('transacciones.csv', index=False)

# Datos de maestro de productos (Excel)
datos_productos = {
    'id_producto': [1, 2, 3, 4],
    'nombre_producto': ['Teclado', 'Mouse', 'Monitor', 'Auriculares'],
    'precio_unitario': [25.50, 15.00, 150.00, 45.00]
}
df_productos_dummy = pd.DataFrame(datos_productos)
df_productos_dummy.to_excel('maestro_productos.xlsx', index=False)

print("¡Archivos de prueba 'transacciones.csv' y 'maestro_productos.xlsx' creados con éxito!")
```

```python

# ==========================================
# 1. ADQUISICIÓN DE DATOS
# ==========================================
print("\n--- Iniciando Adquisición ---")

# Carga del archivo CSV de transacciones
df_transacciones = pd.read_csv('transacciones.csv')

# Carga del archivo Excel del maestro de productos
df_productos = pd.read_excel('maestro_productos.xlsx')

print("Datos cargados correctamente.")
```

```python

# ==========================================
# 2. TRANSFORMACIÓN DE DATOS
# ==========================================
print("\n--- Iniciando Transformación ---")

# Verificar y corregir tipos de datos (Convertir fecha a datetime)
df_transacciones['fecha'] = pd.to_datetime(df_transacciones['fecha'])

# Manejo de valores nulos
# En este contexto, si falta la 'cantidad', imputamos con la mediana (o puedes usar 0, o dropna())
df_transacciones['cantidad'] = df_transacciones['cantidad'].fillna(df_transacciones['cantidad'].median())

# Realizar el merge (combinación) usando 'id_producto' como llave
# Usamos 'left' para mantener todas las transacciones incluso si un producto no existe en el maestro
df_final = pd.merge(df_transacciones, df_productos, on='id_producto', how='left')

print("Transformación completa (Fechas corregidas, nulos manejados y merge realizado).")
```

```python

# ==========================================
# 3. NORMALIZACIÓN / ENRIQUECIMIENTO
# ==========================================
print("\n--- Iniciando Normalización ---")

# Crear la nueva columna 'total_venta'
df_final['total_venta'] = df_final['cantidad'] * df_final['precio_unitario']

print("Columna 'total_venta' calculada con éxito.")
```

```python

# ==========================================
# 4. EXPORTACIÓN
# ==========================================
print("\n--- Iniciando Exportación ---")

# Generar el archivo final en formato Parquet
# Nota: Para esto Pandas requiere tener instalado 'pyarrow' o 'fastparquet' (Colab ya los trae por defecto)
archivo_salida = 'reporte_ventas_final.parquet'
df_final.to_parquet(archivo_salida, index=False)

print(f"¡Proceso terminado! Archivo guardado como '{archivo_salida}'")

# Mostramos una vista previa del resultado final en pantalla
print("\nVista previa del DataFrame final:")
print(df_final.head())
```
